```python
from collections import deque

jug1 = 4
jug2 = 3

queue = deque([(0, 0)])
visited = set()

while queue:
	x, y = queue.popleft()

	if (x, y) in visited:
		continue

	visited.add((x, y))
	print((x, y))

	states = [
		(jug1, 0),
		(0, jug2),
		(x, jug2),
		(jug1, y),
		((x - min(x, jug2 - y)), y + min(x, jug2 - y)),
		((x + min(jug1 - x, y)), y - min(jug1 - x, y))
	]

	for state in states:
		if state not in visited:
			queue.append(state)
			
```
