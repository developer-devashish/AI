```python
def hill_climbing():
    current = 0

    while True:
        current_value = -(current - 5) ** 2 + 25

        right = current + 1
        left = current - 1

        left_value = -(left - 5) ** 2 + 25
        right_value = -(right - 5) ** 2 + 25

        print(f"Current: {current}, Value: {current_value}")

        if left_value > current_value:
            current = left
        elif right_value > current_value:
            current = right
        else:
            break

    print(f"Maximum found at {current}")
    print("Maximum value at ", -(current - 5) ** 2 + 25)
```

```python
hill_climbing()
```
