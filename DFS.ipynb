```python
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F', 'G'],
    'D': [],
    'E': [],
    'F': [],
    'G': []
}

visited = set()
```

```python
def dfs(node):
    if node not in visited:
        print(node, end=" ")
        visited.add(node)

    for neighbour in graph[node]:
        dfs(neighbour)

dfs('A')
```
