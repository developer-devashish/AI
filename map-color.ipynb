```python
def color(node):
    if node == len(regions):
        print(colors)
        return True

    colorss = ["Red", "Green", "Blue"]

    for c in colorss:
        if all(colors[n] != c for n in graph[node]):
            colors[node] = c
            if color(node + 1):
                return True
            colors[node] = None
    return False
```
