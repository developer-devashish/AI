```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 5, 4, 5]
```

```python
n = len(x)
```

```python
sx = sum(x)
sy = sum(y)
sxy = sum(x[i] * y[i] for i in range(n))
sx2 = sum(i * i for i in x)
```

```python
m = (n * sxy - sx * sy) / (n * sx2 - sx * sx)
c = (sy - m * sx) / n
```

```python
print("Slope =", m)
print("Intercept =", c)
print("Equation: y =", m, "x +", c)

# Prediction
x1 = 6
y1 = m * x1 + c
print("Prediction for x =", x1, "is", y1)
```
