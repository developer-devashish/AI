```python
def ab(node, a, b, maxp):
    if isinstance(node, int):
        return node

    if maxp:
        v = -999
        for x in node:
            v = max(v, ab(x, a, b, False))
            a = max(a, v)
            if a >= b: break
        return v

    v = 999
    for x in node:
        v = min(v, ab(x, a, b, True))
        b = min(b, v)
        if a >= b: break
    return v
```

```python
tree = [[3, 5], [6, 9], [1, 2], [0, -1]]
```

```python
print("Best value:", ab(tree, -999, 999, True))
```
