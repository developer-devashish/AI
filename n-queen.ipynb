```python
def is_safe(board, row, col, n):
    for i in range(row):
        if board[i] == col:
            return False

    for i in range(row):
        if board[i] - i == col - row:
            return False

    for i in range(row):
        if board[i] + i == col + row:
            return False

    return True
```

```python
def solve_n_queens(board, row, n):
    if row == n:
        print_board(board, n)
        return True

    for col in range(n):
        if is_safe(board, row, col, n):
            board[row] = col

            if solve_n_queens(board, row + 1, n):
                return True

            board[row] = -1

    return False
```

```python
def print_board(board, n):
    for row in range(n):
        for col in range(n):
            if board[row] == col:
                print("Q", end = " ")
            else:
                print(".", end = " ")
        print()
```

```python
n = int(input("Enter the number of queens: "))
board = [-1] * n

if not solve_n_queens(board, 0, n):
    print("No Solution exists")
```
